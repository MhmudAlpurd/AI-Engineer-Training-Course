```python
X = df.drop('sales', axis=1)
y = df['sales']
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test =  train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
#row 1
#row 2
#row 3
#row 4
#row 5

#test_size = 0.2 ---> 1

#random_state = 42 ---> 2 --> test! ---> mse: 0.1
#random_state = 100 -- > 5 --> test!
```

```python
#y = w1*TV + W2*Radio + w3*Newspaper

from sklearn.linear_model import LinearRegression
model= LinearRegression()

#def LR(X_train, y_train):
        #Gradient Descent!
        #w1 ?
        #w2 ? 
        #w3 ?
        #return w1x1 +  w2x2 +  w3x3
```

```python
X_train.shape
```

```python
#fit ---> GD -->
#y = w1*TV + W2*Radio + w3*Newspaper
# randomly: y = 0.01*TV + 0.3*Radio + 1.5*Newspaper ---> GD
```

```python
model.fit(X_train, y_train) ##OBTAINING WEIGHTS AND BIAS.(W1, W2, W3)
```

```python
model.coef_
```

```python
pd.DataFrame(model.coef_, X.columns, columns=['Weights'])
```

```python
#sales = w1*TV + w2*radio + w3*newspaper
#sales = 0.04*TV + 0.19*radio + 0.006*newspaper
```

```python
# predicting test data / test/ Evaluation
y_pred = model.predict(X_test)
#y_pred = 0.04*TV + 0.19*radio + 0.006*newspaper
```

```python
y_pred
```

```python
X_test
```

```python
#loss functions
from sklearn import metrics
MAE = metrics.mean_absolute_error(y_test, y_pred)
MSE = metrics.mean_squared_error(y_test, y_pred)
RMSE = np.sqrt(MSE)
```

```python
print('mae: ', MAE)
```

```python
print('mse: ', MSE)
```

```python
print('rmse: ', RMSE)
```

```python
test_residual = y_test-y_pred
test_residual
```
