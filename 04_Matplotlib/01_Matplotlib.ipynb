```python
#pip install matplotlib
#from matplotlib import pyplot as plt
import matplotlib.pyplot as plt
```

## Linear diagram

```python
age_x = [25, 26, 27, 28, 29]
dev_y = [2125, 2435, 3000, 3100, 4000]

plt.xlabel('age')
plt.ylabel('salary')
plt.title("developervs avg salary (USD)")
plt.plot(age_x, dev_y)
plt.show()
```

```python
age_x = [25, 26, 27, 28, 29]
py_dev_y = [2125, 2435, 3000, 3100, 4000]
java_dev_y = [2700, 2851, 3210, 3310, 4100]

plt.plot(age_x, py_dev_y, label='py dev salary')
plt.plot(age_x, java_dev_y, label='java dev salary')

plt.xlabel('age')
plt.title('dev salary comparison')
plt.legend()
plt.show()
```

```python
age_x = [25, 26, 27, 28, 29]
py_dev_y = [2125, 2435, 3000, 3100, 4000]
java_dev_y = [2700, 2851, 3210, 3310, 4100]

plt.plot(age_x, py_dev_y, linestyle=':',marker='o', color='b', label='py dev salary')
plt.plot(age_x, java_dev_y, linestyle='--',marker='*', color='g', label='java dev salary')

plt.xlabel('age')
plt.title('dev salary comparison')
plt.legend()
plt.show()
```
