```python
import keras
from keras.datasets import mnist
import matplotlib.pyplot as plt
```

```python
#load 
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
```

```python
train_images.shape
```

```python
test_images.shape
```

```python
train_labels.shape
```

```python
plt.imshow(train_images[8], cmap='gray')
```

```python
train_labels[8]
```

```python
train_labels = keras.utils.to_categorical(train_labels)
test_labels = keras.utils.to_categorical(test_labels)
```

```python
train_labels[8]
```

```python
#Flatten

train_images = train_images.reshape(60000, 784)
test_images = test_images.reshape(10000, 784)
```

```python
train_images.shape
```

```python
test_images.shape
```

```python
train_images[5]
```

```python
train_images[5].shape
```

```python
train_images = train_images.astype('float32')/255.0
test_images = test_images.astype('float32')/255.0
```

```python
train_images[5]
```

```python
from  keras.models import Sequential
from keras.layers import Dense
```

```python
#Design
model = Sequential()
model.add(Dense(128, activation='relu'))#Dense, 128, Relu
model.add(Dense(64, activation='relu'))#Dense, 64, Relu
model.add(Dense(10, activation='softmax'))#Dense, 10, Softmax
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(train_images, train_labels, epochs=5, batch_size=32)
```

```python
model.summary()
```

```python
test_los, test_acc = model.evaluate(test_images, test_labels)
```
